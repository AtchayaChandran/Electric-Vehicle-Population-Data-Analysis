# **Stage 2**

# **Data Cleaning**

```python
# Check Duplicate Rows

duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
# Recheck Missing Values

df.isnull().sum()
```

```python
# Numerical Imputation

numerical_cols = df.select_dtypes(include=np.number).columns

for col in numerical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

print("Numerical missing values imputed successfully")
```

```python
# Categorical Imputation

categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Categorical missing values imputed successfully")
```

```python
# Drop Legislative District

df = df.drop(columns=["Legislative District"])

print("Legislative District column dropped successfully")
```

```python
# Rename Columns

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("-", "_")
)

df.columns
```

```python
# Check Renamed Columns

df.columns
```

```python
# Check Data Types

df.dtypes
```

```python
# Convert Data Types

numeric_columns = [
    "model_year",
    "electric_range",
    "postal_code",
    "legislative_district",
    "dol_vehicle_id",
    "2020_census_tract"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("Data types converted successfully")
```

# **Data Transformation / Data Wrangling**

```python
# Data Transformation - Calculate Vehicle Age

df["vehicle_age"] = 2026 - df["model_year"]

df[["model_year", "vehicle_age"]].head()
```

# **Feature Engineering**

```python
# Feature Engineering - Electric Range Category

df["range_category"] = pd.cut(
    df["electric_range"],
    bins=[-1, 100, 200, 300, 400, float("inf")],
    labels=["0-100", "101-200", "201-300", "301-400", "400+"]
)

print("Range Category created successfully")
```

```python
# Check New Features

df[["model_year", "vehicle_age", "electric_range", "range_category"]].head()
```

# **Statistical Analysis & Visualization**

# **Measure of Central Tendency**

```python
# Numerical Columns for Statistical Analysis

cols = [
    "electric_range",
    "model_year",
    "vehicle_age"
]

# Summary Table

summary_table = pd.DataFrame({
    "Mean": df[cols].mean(),
    "Median": df[cols].median(),
    "Mode": df[cols].mode().iloc[0]
})

print("MEASURE OF CENTRAL TENDENCY\n")
display(summary_table)
```

```python
# Mean, Median and Mode Visualization

for col in cols:

    mean_val = summary_table.loc[col, "Mean"]
    median_val = summary_table.loc[col, "Median"]
    mode_val = summary_table.loc[col, "Mode"]

    plt.figure(figsize=(8, 5))

    sns.histplot(
        df[col],
        kde=True,
        bins=30,
        color="mediumpurple",
        alpha=0.65,
        edgecolor="white"
    )

    plt.axvline(
        mean_val,
        color="crimson",
        linestyle="--",
        linewidth=2.5,
        label=f"Mean: {mean_val:.2f}"
    )

    plt.axvline(
        median_val,
        color="seagreen",
        linestyle="-.",
        linewidth=2.5,
        label=f"Median: {median_val:.2f}"
    )

    plt.axvline(
        mode_val,
        color="darkorange",
        linestyle=":",
        linewidth=3,
        label=f"Mode: {mode_val:.2f}"
    )

    plt.title(
        f"{col.replace('_', ' ').title()} Distribution with Mean, Median & Mode",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(col.replace("_", " "))
    plt.ylabel("Frequency")

    plt.legend(
        loc="upper left",
        bbox_to_anchor=(1, 1)
    )

    plt.grid(axis="y", linestyle="--", alpha=0.25)

    plt.tight_layout()
    plt.show()
```

### **Interpretation – Measures of Central Tendency**

- **Electric Range:** The mean electric range is **37.40 miles**, while the median and mode are both **0.00 miles**. The large difference between the mean and median indicates that the electric range values are unevenly distributed, with some vehicles having considerably higher electric range values. The mode of **0.00 miles** shows that zero is the most frequently occurring value in the dataset.

- **Model Year:** The mean model year is **2022.26**, while both the median and mode are **2023.00**. The closeness of the mean to the median indicates that the model years are mainly concentrated around recent years. The mode of **2023.00** shows that 2023 is the most frequently occurring model year in the dataset.

- **Vehicle Age:** The mean vehicle age is **3.74 years**, while both the median and mode are **3.00 years**. The mean being slightly higher than the median indicates that some older vehicles are increasing the average age. The mode of **3.00 years** shows that 3 years is the most frequently occurring vehicle age in the dataset.

- **Overall:** The measures of central tendency show that the EV dataset contains vehicles with an average electric range of **37.40 miles**, an average model year of **2022.26**, and an average vehicle age of **3.74 years**. The median and mode values further indicate the typical and most frequently occurring values for each variable.

```python
# Variance Visualization - Box Plot

for col in cols:

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        y=df[col],
        color="mediumpurple",
        width=0.45
    )

    plt.title(
        f"{col.replace('_', ' ').title()} - Variance Visualization",
        fontsize=14,
        fontweight="bold"
    )

    plt.ylabel(col.replace("_", " ").title())

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.25
    )

    plt.tight_layout()
    plt.show()
```

### **Interpretation – Variance**

- **Electric Range:** The variance is **5889.14**, indicating a high level of variation in electric range values. This shows that the EVs in the dataset have considerable differences in their electric driving range, with values spread widely around the mean of **37.40 miles**.

- **Model Year:** The variance is **9.49**, which indicates relatively lower variation in model years. This shows that most vehicles are concentrated around recent model years, with the mean model year being **2022.26**.

- **Vehicle Age:** The variance represents the amount of variation in vehicle ages around the mean vehicle age of **3.74 years**. A higher variance would indicate greater differences in vehicle age, while a lower variance would indicate that the vehicles have more similar ages.

- **Overall:** The variance values show that **Electric Range has much greater variability than Model Year**, while Model Year is more concentrated around its average value. This helps understand the spread of the selected numerical variables in the EV dataset.

# **Standard Deviation**

```python
# Standard Deviation Visualization

for col in cols:

    mean_value = df[col].mean()
    std_value = df[col].std()

    plt.figure(figsize=(10, 5))

    sns.histplot(
        df[col],
        bins=30,
        kde=True,
        color="darkorange",
        alpha=0.65,
        edgecolor="white"
    )

    plt.axvline(
        mean_value,
        color="crimson",
        linestyle="--",
        linewidth=2.5,
        label=f"Mean: {mean_value:.2f}"
    )

    plt.axvline(
        mean_value + std_value,
        color="mediumorchid",
        linestyle=":",
        linewidth=2.5,
        label=f"Mean + SD: {mean_value + std_value:.2f}"
    )

    plt.axvline(
        mean_value - std_value,
        color="seagreen",
        linestyle=":",
        linewidth=2.5,
        label=f"Mean - SD: {mean_value - std_value:.2f}"
    )

    plt.title(
        f"{col.replace('_', ' ').title()} - Standard Deviation",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel(col.replace("_", " ").title())
    plt.ylabel("Frequency")

    plt.legend()
    plt.grid(axis="y", linestyle="--", alpha=0.25)

    plt.tight_layout()
    plt.show()
```

### **Interpretation – Standard Deviation**

- **Electric Range:** The standard deviation is **76.74 miles**, which is considerably higher than the mean of **37.40 miles**. This indicates a high level of variation in electric range values, showing that the EVs in the dataset have substantially different driving ranges.

- **Model Year:** The standard deviation is **3.08 years**, with a mean model year of **2022.26**. This relatively small standard deviation indicates that most vehicles are concentrated within a few years around the mean, mainly representing recent model years.

- **Vehicle Age:** The standard deviation is **3.08 years**, with a mean vehicle age of **3.74 years**. This indicates that vehicle ages have some variation around the average, with most vehicles concentrated within a few years of the mean age.

- **Overall:** The standard deviation shows that **Electric Range has the greatest variability**, while **Model Year and Vehicle Age are more concentrated around their respective means**.

**12) How does the distribution of EV types vary across the top counties?**

```python
# Multivariate Analysis - County, EV Type and Vehicle Count

top_counties = df["county"].value_counts().head(10).index

county_data = df[df["county"].isin(top_counties)]

county_ev_type = (
    county_data.groupby(["county", "electric_vehicle_type"])
    .size()
    .reset_index(name="vehicle_count")
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=county_ev_type,
    x="vehicle_count",
    y="county",
    hue="electric_vehicle_type"
)

plt.title("EV Population by County and Electric Vehicle Type")
plt.xlabel("Number of Vehicles")
plt.ylabel("County")
plt.legend(title="EV Type")
plt.show()
```

**Interpretation:**

**This visualization combines County, EV Type, and Vehicle Count. The main metric is the number of registered vehicles for each county and EV type. The chart shows that EV population is concentrated in certain counties, with King County having the highest overall vehicle count, followed by Snohomish and Pierce. Within these counties, the number of BEVs and PHEVs can also be compared. The differences in vehicle counts show that EV adoption is not evenly distributed geographically. Overall, this multivariate analysis helps us understand both the location of EV adoption and the type of electric vehicles present in each county.**
